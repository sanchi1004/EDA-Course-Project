Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Load Dataset

```python
url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/bcancer.csv"
df = pd.read_csv(url)
df.head()
```

Check missing values

```python
df.isnull().sum()
```

```python
sns.heatmap(df.isnull(), cbar=False)
plt.show()
```

Handle missing data

```python
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
df.isnull().sum()
```

Data Cleaning

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.columns = df.columns.str.strip()
```

Data Transformation

```python
df['diagnosis'].value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df['diagnosis'] = le.fit_transform(df['diagnosis'])
```

```python
df.drop(columns=['id', 'Unnamed: 32'], inplace=True)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

feature_cols = df.drop(columns=['diagnosis']).columns
df[feature_cols] = scaler.fit_transform(df[feature_cols])
```

Univariate Analysis - Histogram, Boxplot, Countplot

```python
plt.figure(figsize=(6,4))
plt.hist(df['radius_mean'])
plt.title("Radius Mean")
plt.show()
```

```python
sns.boxplot(x=df['texture_mean'])
plt.show()
```

```python
sns.countplot(x='diagnosis', data=df)
plt.show()
```

Bivariate Analysis - Scatter Plot, Bar Plot, Boxplot

```python
sns.scatterplot(x='radius_mean',
                y='texture_mean',
                hue='diagnosis',
                data=df)
plt.show()
```

```python
sns.barplot(x='diagnosis',
            y='area_mean',
            data=df)
plt.show()
```

```python
sns.boxplot(x='diagnosis',
            y='perimeter_mean',
            data=df)
plt.show()
```

Multivariate Analysis - Heatmap, Pairplot, Violin Plot

```python
plt.figure(figsize=(12,10))
sns.heatmap(df.corr(), cmap='coolwarm')
plt.show()
```

```python
sns.pairplot(df[['radius_mean',
                 'texture_mean',
                 'perimeter_mean',
                 'diagnosis']],
             hue='diagnosis')
```

```python
sns.violinplot(x='diagnosis',
               y='radius_mean',
               data=df)
plt.show()
```
